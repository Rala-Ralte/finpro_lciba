```python
import sys
sys.path.insert(0, '/media/crk/vol3/rt/lciba')

import torch
import matplotlib.pyplot as plt
import numpy as np

from scripts.utils import load_model, load_image, load_text, pil_to_numpy
from scripts.methods import vision_heatmap_libragrad
from scripts.visualization import overlay_vision_heatmap

device = torch.device('cuda')
```

```python
model, preprocess = load_model('ViT-B/32', device)

image_tensor, pil_image = load_image(
    'images/cat.jpg',
    preprocess,
    device
)

text_ids = load_text("a tree branch", device)

# ------------------------------------------------------------
# Use ACTUAL CLIP INPUT image
# ------------------------------------------------------------

image_np = (
    image_tensor[0]
    .detach()
    .cpu()
    .numpy()
)

image_np = image_np.transpose(1,2,0)

image_np = (
    image_np * np.array([
        0.26862954,
        0.26130258,
        0.27577711
    ])
    +
    np.array([
        0.48145466,
        0.4578275,
        0.40821073
    ])
)

image_np = np.clip(
    image_np,
    0,
    1
)

print(next(model.vision_model.parameters()).dtype)
print(image_tensor.dtype)
```

```python
vmaps = {}
for layer_idx in [6, 8, 9, 10]:
    vmaps[layer_idx] = vision_heatmap_libragrad(
        text_ids, image_tensor, model, layer_idx=layer_idx
    )
    print(f"Layer {layer_idx} done")
```

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

for i, layer_idx in enumerate([6, 8, 9, 10]):
    overlaid = overlay_vision_heatmap(image_np, vmaps[layer_idx])
    axes[i].imshow(overlaid)
    axes[i].set_title(f'Layer {layer_idx}')
    axes[i].axis('off')

plt.suptitle('LibraGrad Vision Saliency — Layer Comparison', fontsize=12)
plt.tight_layout()
plt.savefig('outputs/heatmaps/layer_asd.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import torch
from scripts.methods import extract_bert_layer
from scripts.hooks import ActivationGradientRecorder

layer = extract_bert_layer(model.vision_model, 9)

captured = {}
def fwd_hook(m, inp, out):
    o = out[0] if isinstance(out, tuple) else out
    o.retain_grad()
    captured['out'] = o

h = layer.register_forward_hook(fwd_hook)

model.zero_grad()
img_feat = model.get_image_features(image_tensor)
txt_feat = model.get_text_features(text_ids)
loss = -torch.nn.CosineSimilarity(dim=-1)(img_feat, txt_feat).mean()
loss.backward()

h.remove()

act  = captured['out']
grad = captured['out'].grad

patch_sal_layer9 = (act * grad).abs().sum(-1)[0, 1:].detach().cpu().numpy()
print(f"patch_sal_layer9 shape: {patch_sal_layer9.shape}")
print(f"max at index: {patch_sal_layer9.argmax()} (row={patch_sal_layer9.argmax()//7}, col={patch_sal_layer9.argmax()%7})")
model.zero_grad()
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import torch.nn.functional as F
from scripts.utils import normalize

# Test 1: top-left hotspot
grid_tl = np.zeros((7,7), dtype=np.float32)
grid_tl[0,0] = 1.0

# Test 2: bottom-right hotspot  
grid_br = np.zeros((7,7), dtype=np.float32)
grid_br[6,6] = 1.0

fig, axes = plt.subplots(1, 4, figsize=(16, 4))

for i, (grid, title) in enumerate([
    (grid_tl, 'Top-left [0,0]=1'),
    (grid_br, 'Bottom-right [6,6]=1'),
]):
    t = torch.tensor(grid).reshape(1,1,7,7)
    up = F.interpolate(t, size=224, mode='bilinear', align_corners=False)[0,0].numpy()
    
    axes[i].imshow(up, cmap='jet')
    axes[i].set_title(title)

# Test 3: overlay on cat using _reverse_clip_normalization
from scripts.visualization import _reverse_clip_normalization

# image_tensor is CLIP normalized - reverse it
img_display = _reverse_clip_normalization(image_tensor[0].cpu().numpy())

# Fresh vmap with F.interpolate
patch_sal = patch_sal_layer9  # use the retained variable from earlier
grid = torch.tensor(patch_sal).reshape(1,1,7,7).float()
vmap_correct = F.interpolate(grid, size=224, mode='bilinear', align_corners=False)[0,0].numpy()
vmap_correct = normalize(vmap_correct)

axes[2].imshow(img_display)
axes[2].imshow(vmap_correct, cmap='jet', alpha=0.5)
axes[2].set_title('Overlay on reversed CLIP image')

axes[3].imshow(image_np)
axes[3].imshow(vmap_correct, cmap='jet', alpha=0.5)
axes[3].set_title('Overlay on pil image_np')

plt.tight_layout()
plt.savefig('outputs/heatmaps/diagnostic_definsitive.png', dpi=150, bbox_inches='tight')
plt.show()
```
