```python
from scripts.utils import load_model, load_text, load_image
from scripts.hooks import install_libragrad_hooks, remove_libragrad_hooks, verify_hooks_dont_change_forward
import clip, torch

model, preprocess = load_model("ViT-B/32")
```

```python
dummy_img = torch.randn(1, 3, 224, 224).cuda()
dummy_txt = clip.tokenize(["a cat on a table"]).cuda()
verify_hooks_dont_change_forward(model, dummy_img, dummy_txt)
```

```python
from scripts.libragrad import compute_libragrad_vision_saliency
img_t, pil = load_image("images/cat.jpg", preprocess)
txt_t = load_text("a cat sitting on a table")
target_layer = model.vision_model.transformer.resblocks[9]

vmap = compute_libragrad_vision_saliency(model, img_t, txt_t, target_layer)
print(f"vmap: min={vmap.min():.4f} max={vmap.max():.4f} std={vmap.std():.4f}")

import matplotlib.pyplot as plt
plt.imshow(vmap, cmap='jet'); plt.colorbar(); plt.title("LibraGrad vision saliency"); plt.show()
```

```python
from scripts.libragrad import prepare_vision_prior
from scripts.methods import extract_feature_map

features = extract_feature_map(model.vision_model, 9, img_t)
prior = prepare_vision_prior(vmap, hidden_dim=features.shape[-1])
print(f"prior: shape={prior.shape} min={prior.min():.4f} max={prior.max():.4f} std={prior.std():.4f}")
```

```python
from scripts.iba import InformationBottleneck
from scripts.methods import get_compression_estimator, extract_bert_layer

layer = extract_bert_layer(model.vision_model, 9)
estimator = get_compression_estimator(1.0, layer, features)
bn = InformationBottleneck(estimator.mean(), estimator.std(), device="cuda")

print(f"alpha default: min={bn.alpha.min():.4f} max={bn.alpha.max():.4f} std={bn.alpha.std():.4f}")
bn.libragrad_initialize_alpha(prior, strength=3.0)
print(f"alpha after init: min={bn.alpha.min():.4f} max={bn.alpha.max():.4f} std={bn.alpha.std():.4f}")
```

```python
from scripts.methods import vision_heatmap_lciba, vision_heatmap_iba

kwargs = dict(layer_idx=9, beta=0.1, var=1.0, lr=1.0, train_steps=10, progbar=True)

# M2IB first (as compare_methods.py does)
vmap_m2ib  = vision_heatmap_iba(txt_t, img_t, model, **kwargs)

# Then LC-IBA (same model, same order as compare_methods.py)
vmap_lciba = vision_heatmap_lciba(txt_t, img_t, model, **kwargs)

print(f"M2IB  vmap std: {vmap_m2ib.std():.4f}")
print(f"LC-IBA vmap std: {vmap_lciba.std():.4f}")

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(vmap_m2ib,  cmap='jet'); axes[0].set_title("M2IB")
axes[1].imshow(vmap_lciba, cmap='jet'); axes[1].set_title("LC-IBA")
plt.show()
```

```python
import gc

# Free existing model
del model
torch.cuda.empty_cache()
gc.collect()

# Load fresh — no M2IB before it
model_fresh, _ = load_model("ViT-B/32")

vmap_lciba_solo = vision_heatmap_lciba(txt_t, img_t, model_fresh, **kwargs)
print(f"LC-IBA solo  vmap std: {vmap_lciba_solo.std():.4f}")
print(f"LC-IBA after M2IB std: {vmap_lciba.std():.4f}")   # from Cell 6

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(vmap_lciba_solo, cmap='jet'); axes[0].set_title("LC-IBA solo (fresh model)")
axes[1].imshow(vmap_lciba,      cmap='jet'); axes[1].set_title("LC-IBA after M2IB")
plt.show()
```
